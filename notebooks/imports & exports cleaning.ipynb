## ***Importing Libraries***

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

## ***Data Loading & Overview:***

### 1. **Loading the Dataset**

```python
def load_data(file_path):
    """ load the dataset from excel file and return dataframe """
    try:
        data = pd.read_excel(file_path)
    except Exception as e:
        print(f"file not found error: {e}")
        return None
    else:
        print("data loaded successfully")
        return data
    finally:
        print("Execution completed.")

file_path = "../Data/Raw Data/imports and exports.xlsx"
df = load_data(file_path)
df.head()
```

### 2. ***Data Overview***

```python
# data shape
print(f"Data shape: {df.shape[0]} rows and {df.shape[1]} columns")

#data info & data types
df.info()
```

```python
numeric_cols = ['Imports', 'Exports','Imports Value', 'Exports Value']

print("Statistical Analysis:")
df[numeric_cols].describe().T
```

```python
print("\nStatistical Summary for Categorical Variables:")
df.describe(include=object).T
```

## ***Data Quality Check & Cleaning***

### 1. ***Data Quality Check***

#### **Duplicated Rows**

```python
print("Full duplicates:", df.duplicated().sum())
print("Duplicate business keys:", 
    df.duplicated(
        subset=['Crop','Year']).sum())
```

#### **Missing Values**

```python
missing_values = pd.DataFrame({
    'Missing_Count': df.isnull().sum(),
    'Missing_Percentage': (df.isnull().sum() / len(df)) * 100})

missing_values
```

```python
missing_data = df[df[['Imports', 'Exports','Imports Value', 'Exports Value']].isnull().any(axis=1)]

missing_data.sort_values(
    by=['Crop','Year'])
```

```python
print("\nNegative Values:")
print((df[numeric_cols] < 0).sum())
```

```python
print("\nZero Values:")
print((df[numeric_cols] == 0).sum())
```

```python
print("\nZero values distribution across Capita:")
print(df[df['Imports'] == 0]['Crop'].value_counts())
```

```python
print("\nUnique Crops Count:", df['Crop'].nunique())
print("Crops List:", df['Crop'].unique())
print("Years Range:", df['Year'].min(), "to", df['Year'].max())
```

### 2. ***Data Cleaning***

```python
df_cleaned = df.copy()
```

```python
df_cleaned.columns = (
    df_cleaned.columns.str.strip()
    .str.replace(" ","_", regex=False)
    .str.replace("-","_", regex=False)
    .str.replace("/","_", regex=False)
    .str.lower())

print(df_cleaned.columns)
```

```python
df_cleaned.to_csv('../Data/Clean Data/cleaned_imports_exports.csv', index=False, encoding='utf-8-sig')

print("CSV file saved successfully!")
print("Shape:", df_cleaned.shape)
```
